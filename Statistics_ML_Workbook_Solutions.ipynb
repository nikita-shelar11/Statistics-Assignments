## Part 1 — Measures of Central Tendency (Q1–Q15)

### Q1. Arithmetic Mean & Deviations

```python
sales = np.array([48, 52, 61, 57, 49, 63, 70])   # data
mean = sales.mean()                              # arithmetic mean = sum/n
deviations = sales - mean                         # each value minus the mean
print("Mean:", mean)
print("Deviations:", deviations)
print("Sum of deviations (should be ~0):", deviations.sum())   # verifies property
```

### Q2. Mean with Missing Observation

```python
n, mean = 6, 38
known = np.array([24, 35, 42, 40, 45])   # 5 known numbers
total = n * mean                          # sum of all 6 numbers = n * mean
missing = total - known.sum()             # remaining value
print("Missing number:", missing)
```

### Q3. Effect of a New Observation

```python
old_n, old_mean, new_val = 8, 27, 45
new_mean = (old_n*old_mean + new_val) / (old_n+1)   # add new total, divide by new n
print("New mean:", new_mean)
```

### Q4. Combined Mean of Two Groups

```python
n1, m1 = 35, 68
n2, m2 = 25, 74
combined_mean = (n1*m1 + n2*m2) / (n1+n2)   # weighted by group sizes
print("Combined mean:", combined_mean)
```

### Q5. Weighted Mean

```python
scores  = np.array([82, 76, 88])     # assignment, midterm, final
weights = np.array([0.20, 0.30, 0.50])
weighted_mean = np.sum(scores*weights) / weights.sum()   # sum(w*x)/sum(w)
print("Weighted mean:", weighted_mean)
```

### Q6. Median (Odd n)

```python
temps = np.array([29, 31, 28, 35, 33, 30, 32])
median = np.median(temps)      # middle value once sorted (odd count -> single middle)
print("Sorted:", np.sort(temps))
print("Median:", median)
```

### Q7. Median (Even n)

```python
expenses = np.array([2200, 2500, 1800, 3100, 2900, 2600, 2400, 2800])
median = np.median(expenses)   # average of the two middle values (even count)
print("Sorted:", np.sort(expenses))
print("Median:", median)
```

### Q8. Mode

```python
from collections import Counter
data = [12, 15, 18, 15, 20, 18, 15, 22, 18, 15]
counts = Counter(data)                     # frequency of each value
max_freq = max(counts.values())
modes = [k for k,v in counts.items() if v == max_freq]   # all values tied for highest freq
print("Frequencies:", counts)
print("Mode(s):", modes, "-> unimodal" if len(modes)==1 else "-> multimodal")
```

### Q9. Mean, Median, Mode together

```python
data = np.array([15, 16, 18, 18, 19, 20, 20, 21, 24])
mean = data.mean()
median = np.median(data)
mode = stats.mode(data, keepdims=True).mode[0]   # most frequent value
print(f"Mean={mean:.2f}, Median={median}, Mode={mode}")
# Comment: mean > median > ... roughly close together -> distribution nearly symmetric
```

### Q10. Effect of an Outlier

```python
data = np.array([18, 20, 21, 22, 23, 24, 90])   # 90 is an outlier
mean = data.mean()
median = np.median(data)
print(f"Mean={mean:.2f}, Median={median}")
# Median is more robust/representative here: the outlier (90) drags the mean up,
# while the median barely changes since it only depends on rank/position.
```

## Part 2 — Measures of Dispersion (Q16–Q30)

### Q16. Range & Coefficient of Range

```python
prod = np.array([125, 138, 142, 119, 151, 136, 145, 132])
rng = prod.max() - prod.min()                        # range = max - min
coeff_range = rng / (prod.max()+prod.min())           # (max-min)/(max+min)
print("Range:", rng, "| Coefficient of Range:", round(coeff_range, 4))
```

### Q17. Quartiles & IQR

```python
scores = np.array([42,45,48,50,53,55,57,60,62,64,66,68,70,72,75])
Q1 = np.percentile(scores, 25)     # 25th percentile
Q3 = np.percentile(scores, 75)     # 75th percentile
IQR = Q3 - Q1                       # spread of the middle 50%
QD = IQR / 2                        # quartile deviation (semi-IQR)
print(f"Q1={Q1}, Q3={Q3}, IQR={IQR}, QD={QD}")
```

### Q18. Population Variance & SD

```python
defects = np.array([3, 5, 4, 6, 7, 5])
pop_mean = defects.mean()
pop_var  = defects.var(ddof=0)     # ddof=0 -> divide by N (population)
pop_sd   = defects.std(ddof=0)
print(f"Mean={pop_mean}, PopVar={pop_var}, PopSD={pop_sd:.3f}")
```

### Q19. Sample Variance & SD

```python
weights = np.array([18, 22, 21, 20, 19, 23, 17])
sample_mean = weights.mean()
sample_var  = weights.var(ddof=1)   # ddof=1 -> divide by (n-1), unbiased sample variance
sample_sd   = weights.std(ddof=1)
print(f"Mean={sample_mean}, SampleVar={sample_var:.3f}, SampleSD={sample_sd:.3f}")
```

### Q20. Comparing Variability of Two Machines

```python
A = np.array([48, 50, 49, 51, 52])
B = np.array([42, 46, 50, 54, 58])
for name, arr in [("A", A), ("B", B)]:
    print(f"Machine {name}: mean={arr.mean()}, sd={arr.std(ddof=0):.3f}")
# Lower SD = more consistent lengths -> Machine A is more consistent (smaller spread)
```

### Q21. Coefficient of Variation

```python
portfolios = {"A": (12, 2), "B": (18, 4)}   # (mean%, sd%)
for name, (mean, sd) in portfolios.items():
    cv = sd/mean*100                          # CV = (SD/Mean)*100, unit-free measure of risk
    print(f"Portfolio {name}: CV = {cv:.2f}%")
# Lower CV = more consistent per unit of return -> compare printed values
```

### Q22. Effect of Adding a Constant

```python
data = np.array([14, 18, 22, 26, 30])
mean0, var0, sd0 = data.mean(), data.var(ddof=0), data.std(ddof=0)
new_mean = mean0 + 8      # adding constant c shifts the mean by c
new_var  = var0           # variance is UNCHANGED by adding a constant (spread doesn't change)
new_sd   = sd0            # SD unchanged for same reason
print(f"New mean={new_mean}, New variance={new_var}, New SD={new_sd}")
```

### Q23. Effect of Multiplying by a Constant

```python
data = np.array([5, 8, 11, 14, 17])
mean0, var0, sd0 = data.mean(), data.var(ddof=0), data.std(ddof=0)
k = 4
new_mean = mean0 * k       # mean scales by k
new_var  = var0 * k**2     # variance scales by k^2
new_sd   = sd0 * k         # SD scales by k
print(f"New mean={new_mean}, New variance={new_var}, New SD={new_sd}")
```

### Q24. Z-Scores

```python
mu, sigma = 68, 6
for score in [74, 60, 68]:
    z = (score - mu) / sigma          # z = (x - mean) / sd
    print(f"Score {score}: z = {z:.2f}")
# z=1 -> 1 SD above mean; z=-1.33 -> below mean; z=0 -> exactly at the mean
```

### Q26. Variance from a Frequency Distribution
*(Placeholder table — replace with your image's actual customer-count data)*

```python
values = np.array([0, 1, 2, 3, 4])          # REPLACE with actual customer counts
freq   = np.array([10, 20, 30, 25, 15])     # REPLACE with actual frequencies
mean = np.sum(values*freq)/freq.sum()
var  = np.sum(freq*(values-mean)**2)/freq.sum()   # population variance for grouped data
sd = math.sqrt(var)
print(f"Mean={mean:.2f}, Population Variance={var:.2f}, Population SD={sd:.2f}")
```

### Q27. Detecting Outliers Using IQR

```python
incomes = np.array([320,340,350,360,365,370,372,375,380,385,390,520])
Q1, Q3 = np.percentile(incomes, [25, 75])
IQR = Q3 - Q1
lower_fence = Q1 - 1.5*IQR     # standard Tukey fences
upper_fence = Q3 + 1.5*IQR
outliers = incomes[(incomes < lower_fence) | (incomes > upper_fence)]
print(f"Q1={Q1}, Q3={Q3}, IQR={IQR}")
print(f"Fences: [{lower_fence}, {upper_fence}]")
print("Outlier(s):", outliers)
```

### Q29. Missing Observation (variance = 16)

```python
import sympy as sp
x = sp.symbols('x', real=True)
data = [6, 8, 10, x, 14]
n = len(data)
mean_expr = sum(data)/n
var_expr = sum((d-mean_expr)**2 for d in data)/n     # population variance formula
solutions = sp.solve(sp.Eq(sp.simplify(var_expr), 16), x)
print("x =", solutions)
```

### Q41. Total Probability (Factory Machines)

```python
pA, pB, pC = 0.40, 0.35, 0.25
dA, dB, dC = 0.02, 0.03, 0.05
p_defective = pA*dA + pB*dB + pC*dC     # law of total probability
print("P(defective) =", p_defective)
```

### Q42. Bayes' Theorem (Machine C given defective)

```python
p_defective = pA*dA + pB*dB + pC*dC
p_C_given_defective = (pC*dC) / p_defective   # Bayes: P(C|D) = P(D|C)P(C) / P(D)
print("P(Machine C | defective) =", round(p_C_given_defective, 4))
```

### Q43. Union & Conditional Probability

### Q47. Bayes' Theorem — Diagnostic Test

```python
p_disease   = 0.02
sensitivity = 0.95              # P(positive | disease)
specificity = 0.98              # P(negative | no disease)
p_pos_given_disease    = sensitivity
p_pos_given_no_disease = 1 - specificity
p_positive = p_pos_given_disease*p_disease + p_pos_given_no_disease*(1-p_disease)
p_disease_given_positive = (p_pos_given_disease*p_disease) / p_positive
print("P(Disease | Positive) =", round(p_disease_given_positive, 4))
```

## Part 6 — Hypothesis Testing (Q96–Q100)
General Z-test recipe used below:
1. State H0 (no difference / claim holds) vs H1 (claim is false).
2. Compute `Z = (sample_mean - mu0) / (sigma/sqrt(n))`.
3. Compare `Z` to the critical value (or use the p-value) to decide.

### Q96. Two-Tailed Z-Test — Packet Weight

```python
mu0, sigma, n, xbar, alpha = 500, 20, 64, 495, 0.05
se = sigma/math.sqrt(n)
Z = (xbar - mu0)/se                                  # test statistic
p_value = 2*(1 - stats.norm.cdf(abs(Z)))             # two-tailed p-value
z_crit = stats.norm.ppf(1-alpha/2)                   # +-1.96 for alpha=0.05

print("H0: mu = 500 g   |   H1: mu != 500 g")
print(f"Z = {Z:.3f}, critical Z = +-{z_crit:.3f}, p-value = {p_value:.4f}")
print("Reject H0" if abs(Z) > z_crit else "Fail to reject H0")
```

### Q97. Two-Tailed Z-Test — Electricity Bill

```python
mu0, sigma, n, xbar, alpha = 2500, 300, 100, 2420, 0.01
se = sigma/math.sqrt(n)
Z = (xbar - mu0)/se
p_value = 2*(1 - stats.norm.cdf(abs(Z)))
z_crit = stats.norm.ppf(1-alpha/2)

print(f"Z = {Z:.3f}, critical Z = +-{z_crit:.3f}, p-value = {p_value:.5f}")
print("Reject H0 (claim of ₹2500 is not supported)" if abs(Z) > z_crit else "Fail to reject H0")
```

### Q98. One-Tailed Z-Test — Battery Life (claim: exceeds 15 hrs)

```python
mu0, sigma, n, xbar, alpha = 15, 2.8, 49, 15.8, 0.05
se = sigma/math.sqrt(n)
Z = (xbar - mu0)/se
p_value = 1 - stats.norm.cdf(Z)          # right-tailed test since H1: mu > 15
z_crit = stats.norm.ppf(1-alpha)

print("H0: mu <= 15   |   H1: mu > 15")
print(f"Z = {Z:.3f}, critical Z = {z_crit:.3f}, p-value = {p_value:.4f}")
print("Reject H0 (evidence battery life exceeds 15 hrs)" if Z > z_crit else "Fail to reject H0")
```

### Q99. Two-Tailed Z-Test — Student Scores

```python
mu0, sigma, n, xbar, alpha = 70, 9, 81, 72, 0.05
se = sigma/math.sqrt(n)
Z = (xbar - mu0)/se
p_value = 2*(1 - stats.norm.cdf(abs(Z)))
z_crit = stats.norm.ppf(1-alpha/2)

print(f"Z = {Z:.3f}, critical Z = +-{z_crit:.3f}, p-value = {p_value:.4f}")
print("Reject H0 (mean differs from 70)" if abs(Z) > z_crit else "Fail to reject H0")
```

### Q100. Full Z-Test + Confidence Interval — Bottle Filling

```python
mu0, sigma, n, xbar, alpha = 1.0, 0.04, 100, 0.985, 0.05
se = sigma/math.sqrt(n)

Z = (xbar - mu0)/se                              # 1. test statistic
p_value = 2*(1 - stats.norm.cdf(abs(Z)))         # 2. two-tailed p-value
z_crit = stats.norm.ppf(1-alpha/2)               # 3. critical value (~1.96)
reject = abs(Z) > z_crit                         # 4. decision

margin = z_crit*se                               # 5. 95% CI: xbar +- z*SE
ci_low, ci_high = xbar-margin, xbar+margin

print("H0: mu = 1.0 L   |   H1: mu != 1.0 L")
print(f"Z = {Z:.3f}, p-value = {p_value:.5f}, critical Z = +-{z_crit:.3f}")
print("Reject H0" if reject else "Fail to reject H0")
print(f"95% CI for mu: ({ci_low:.4f}, {ci_high:.4f}) litres")
# 6. Interpretation: since 1.0 L is NOT inside the CI, this confirms the machine
#    is under-filling bottles on average -> supports rejecting H0.
```
